### Aufgabe 1: Relative und globale Phase

In der ersten Aufgabe untersuchen wir den Unterschied zwischen der
relativen und der globalen Phase eines Quantenzustands.

Eine **globale Phase** betrifft alle Amplituden gleichermaßen. Zum Beispiel
unterscheiden sich $\ket{\Psi}$ und $-\ket{\Psi}$ nur durch ein gemeinsames
Minuszeichen. Das ändert keine Messwahrscheinlichkeiten.

Eine **relative Phase** betrifft die Phasen der Amplituden zueinander.
Zum Beispiel unterscheiden sich
$\ket{\Psi_1} = \frac{1}{\sqrt{2}}(\ket{0}+\ket{1})$ und
$\ket{\Psi_2} = \frac{1}{\sqrt{2}}(\ket{0}-\ket{1})$ durch das Vorzeichen einer Amplitude.
Diese Zustände lassen sich durch geeignete Gatter und anschließende Messung
unterscheiden.

### Aufgabe 3: Messungen

Bisher haben wir den Zustand und die theoretischen Messwahrscheinlichkeiten
mit einem Simulator berechnet. Ein einzelnes Messergebnis verrät uns diese
Wahrscheinlichkeiten noch nicht. Wir schätzen sie, indem wir den gleichen
Schaltkreis oft vorbereiten und messen.

In Qiskit speichern **klassische Bits** die Messergebnisse. Bei
`QuantumCircuit(num_qubits, num_bits)` gibt das erste Argument die Anzahl
der Qubits und das zweite die Anzahl der klassischen Bits an.

Mit `quantum_circuit.measure(0, 0)` messen wir Qubit 0 und speichern das
Ergebnis im klassischen Bit 0. Der erste Index bezeichnet also das Qubit,
der zweite das klassische Bit.

```python
# Ein Qubit und ein klassisches Bit
quantum_circuit = QuantumCircuit(1, 1)
quantum_circuit.h(0)
quantum_circuit.measure(0, 0)
quantum_circuit.draw("mpl")
```

Im Schaltbild bezeichnet `c` das klassische Register für die Messergebnisse.
Die doppelte Linie steht für klassische Information: Hier werden die Werte
0 oder 1 gespeichert.

#### Simulierte Messungen mit Qiskit

Ein **Shot** ist eine Ausführung des Schaltkreises einschließlich der Messung.
Für jeden Shot werden die Qubits neu im Anfangszustand vorbereitet und alle
Gatter erneut angewendet. Wir messen also nicht immer wieder denselben,
bereits gemessenen Zustand.

Die Hilfsfunktion `sample_from_circuit(quantum_circuit, num_shots=1024)`
führt solche Messungen auf einem Simulator aus. `num_shots` muss eine
positive ganze Zahl sein.

Das Ergebnis ist ein Dictionary mit den Häufigkeiten der gemessenen
Bitstrings. Zum Beispiel bedeutet `{"00": 15, "01": 20, "10": 35, "11": 30}`,
dass bei insgesamt 100 Shots der Bitstring `"00"` genau 15-mal gemessen wurde.
Ein Bitstring, der nicht gemessen wurde, kann im Dictionary fehlen.

```python
from qc_lecture_tools.sampling import sample_from_circuit

# Beispiel: Nur das erste Qubit wird in Superposition gebracht.
quantum_circuit = QuantumCircuit(2, 2)
quantum_circuit.h(0)

display(sv_state(quantum_circuit, "Quantenzustand vor der Messung: "))
display(sv_probs(quantum_circuit, "Theoretische Messwahrscheinlichkeiten: "))

quantum_circuit.measure(0, 0)
quantum_circuit.measure(1, 1)

sample_from_circuit(quantum_circuit, num_shots=1000)
```
