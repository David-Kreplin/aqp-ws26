### Berechnung von Zustandsvektoren und Quantenzuständen aus Quantenschaltungen



Qiskit kann den resultierenden Quantenzustand einer Quantenschaltung berechnen.
Zur Erinnerung: Der Quantenzustand eines Qubits kann in zwei äquivalenten Schreibweisen dargestellt werden ($\alpha, \beta$ sind Amplituden):

- Zustandsvektor: $\ket{\psi} = \begin{pmatrix} \alpha \\ \beta \end{pmatrix}$
- Quantenzustand (Dirac-Notation): $\ket{\psi} = \alpha \ket{0} + \beta \ket{1}$

Um beide Ausdrücke für eine gegebene Quantenschaltung zu bestimmen, verwenden wir die folgenden Funktionen aus der Hilfsbibliothek ``qc_lecture_tools``:

> - ``sv_vector(quantum_circuit, label=None)``: zeigt den resultierenden **Zustandsvektor** einer Quantenschaltung an
> 
> - ``sv_state(quantum_circuit, label=None)``: zeigt den resultierenden **Quantenzustand** einer Quantenschaltung an
> 
> - ``sv_probs(quantum_circuit, label=None)``: zeigt die resultierenden **Messwahrscheinlichkeiten** nach Anwendung der Quantenschaltung an

Alle Funktionen haben ein optionales zweites Argument für eine Beschriftung. Mehrere Darstellungen können mit ``display()`` angezeigt werden.

```python
from qc_lecture_tools.statevector import sv_vector, sv_state, sv_probs  # Importiere Hilfsfunktionen
```

```python
# Sie können dieses Code-Snippet im gesamten Lab kopieren, um den Zustandsvektor,
# den Quantenzustand und die Messwahrscheinlichkeiten Ihrer Quantenschaltung zu erhalten.

# Beispiel: Hadamard-Gatter
quantum_circuit = QuantumCircuit(1)
quantum_circuit.h(0)

# Ausgabe des Zustandsvektors, des Quantenzustands und der Messwahrscheinlichkeiten
display(sv_vector(quantum_circuit, "Zustandsvektor: "))
display(sv_state(quantum_circuit, "Quantenzustand: "))
display(sv_probs(quantum_circuit, "Messwahrscheinlichkeiten: "))
```
