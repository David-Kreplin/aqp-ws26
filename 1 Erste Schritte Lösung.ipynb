Willkommen zum ersten Praktikum.
Wir arbeiten nur mit sehr kleinen Quantenschaltkreisen und beobachten, was bei einzelnen Messungen passiert .

Zur Erinnerung:

- Ein **Qubit** ist das quantenmechanische Analogon zu einem Bit.
- Bei einer Messung erhalten wir aber immer nur klassische Ergebnisse, also `0` oder `1`.
- Bei Quantenzuständen schreibe nutzen wir die Eckigen Klammern (Kets), also $\ket{0}$ und $\ket{1}$.
- Einige Quantengatter verändern den Zustand deterministisch, andere erzeugen Überlagerungen, bei denen Messungen zufällige Ergebnisse liefern können.
- Eine Messung zerstört die überlagerung im Quantenzustand, und der Quantenzustand *kollabiert* in den gemessenen Zustand.  

Wir benutzen in diesem Praktikum **Qiskit**, um Quantenschaltkreise zu erzeugen und zu visualisieren.

````python
from qiskit import QuantumCircuit
from qc_lecture_tools.sampling import measure_quantum_circuit
````

Ein Quantenschaltkreis wird mit der Klasse `QuantumCircuit` erstellt.
Die Zahl im Argument gibt an, wie viele Qubits der Schaltkreis enthält.

Beispiel für einen leeren Schaltkreis mit einem Qubit:

````python
num_qubits = 1
quantum_circuit = QuantumCircuit(num_qubits)
quantum_circuit.draw("mpl")
````

Quantengatter werden als Methoden an den Schaltkreis angehängt, sie werden einfach nacheinander Aufgerufen. Als Argument wird der Index des gewünschten Qubits übergeben. Bei einem zwei-Qubit Quantengatter werden das Kontroll- und Zielqubit übergeben. 

Beispiele:

- `quantum_circuit.x(0)` fügt ein X-Gatter (auch Not-Gatter gennant) auf Qubit 0 ein.
- `quantum_circuit.h(0)` fügt ein Hadamard-Gatter auf Qubit 0 ein.
- `quantum_circuit.cx(0, 1)` fügt ein CNOT-Gatter mit Kontrollqubit 0 und Zielqubit 1 ein.

**Wichtig:** In Python beginnt die Zählung bei 0. Das erste Qubit hat also den Index 0.

#### Zwischenbemerkung

Bis hierhin sieht man bereits einen wichtigen Unterschied zwischen klassischen und quantenmechanischen Operationen:

- Das erste Qubit ist das **Kontrollqubit**.

In Qiskit schreiben wir dafür zum Beispiel:


**Bonusaufgabe:**

Wenn Sie bereits etwas mehr Erfahrung mit Python haben, können Sie die beobachteten Zustände auch automatisch zählen.

Nutzen Sie dafür bei der Messfunktion die Option ``latex=False``:

>```python
>measure_quantum_circuit(quantum_circuit, latex=False)
>```

Dann wird der gemessene Zustand nicht als formatierte Ausgabe, sondern als String zurückgegeben, zum Beispiel ``"10"``.
Diesen Rückgabewert können Sie anschließend in Python weiterverarbeiten.
Führen Sie die Messung dazu mehrfach hintereinander aus und verwenden Sie zum Beispiel ein Python-Dictionary (oder eine andere geeignete Datenstruktur), um zu zählen, wie oft jeder gemessene Zustand auftritt.
Auf diese Weise können Sie die Häufigkeiten der Messergebnisse systematisch untersuchen.

Testen sie verschieden Gatterkombinationen aus den vorherigen Aufgaben aus.  

````python
def count_measurements(quantum_circuit: QuantumCircuit, num_measurements:int) -> dict:
    """ 
    Function to count the measurements of a quantum circuit

    Args:
        quantum_circuit (QuantumCircuit): The quantum circuit to measure
        num_measurements (int): The number of measurements to perform

    Returns:
        dict: A dictionary with the counts of the measurements
    """
    
    # Empty dictionary to store the counts
    counts = {}
    for _ in range(num_measurements):
        # Get the bitstring from the measurement
        bitstring = measure_quantum_circuit(quantum_circuit, latex=False)
        # Update the counts in the dictionary
        if bitstring in counts:
            counts[bitstring] += 1
        else:
            counts[bitstring] = 1
    return counts
````

````python
# Messungen aus Aufgabe 2.3 zählen
quantum_circuit = QuantumCircuit(1)
quantum_circuit.h(0)
counts = count_measurements(quantum_circuit, 1000)
print(counts)
````

````python
# Messungen aus Aufgabe 3.1 zählen
quantum_circuit = QuantumCircuit(2)
quantum_circuit.h(0)
quantum_circuit.cx(0,1)
counts = count_measurements(quantum_circuit, 1000)
print(counts)
````
